```python
tensor = torch.randint(1, 9, (4, 6))
tril = torch.tril(tensor)

pprint(tensor)
pprint(tril)
```

```python
tensor = torch.randint(1, 9, (4, 6))
tril = torch.tril(tensor, diagonal=2)

pprint(tensor)
pprint(tril)
```

Аргумент diagonal позволяет управлять выбором линии диагонали

#### **Tensor Views**

PyTorch позволяет тензору быть представлением (view) существующего тензора

Представление тензора использует те же базовые данные, что и базовый тензор.
Поддержка View позволяет избежать явного копирования данных, что дает возможность выполнять быстрое и эффективное с точки зрения использования памяти изменение формы, нарезку и поэлементные операции.

При создании представления данные не перемещаются, тензор представления просто по-новому интерпретирует те же данные.

```python
t = torch.rand(4, 4)
b = t.view(2, 8)
```

```python
pprint(t)
pprint(b)
```

```python
tensor = torch.randint(0, 5, (1, 1, 2, 3))
sq_tensor = torch.squeeze(tensor)

pprint(tensor)
pprint(sq_tensor)
print(tensor.shape, sq_tensor.shape)
```

`squeeze` возвращает тензор, в котором удалены все измерения input размером 1.

---

```python
tensor = torch.randint(0, 5, (5, ))
un1_tensor = torch.unsqueeze(tensor, dim=0)
un2_tensor = torch.unsqueeze(tensor, dim=1)

pprint(tensor)
pprint(un1_tensor)
pprint(un2_tensor)
print(tensor.shape, un1_tensor.shape, un2_tensor.shape)
```

`unsqueeze` - возвращает новый тензор с измерением размером один, вставленным в указанную позицию.

```python
tensor = torch.randint(0, 5, (2, 3, 4))
pprint(tensor)
```

```python
pprint(torch.flatten(tensor))
pprint(torch.flatten(tensor, start_dim=1))
```

`flatten` сплющивает тензор (преобразовывает в плоский тензор). Не всегда возвращает view

`ravel` производит ту же операцию. Но при этом старается вернуть представление (view), если это возможно

---

```python
tensor = torch.randint(0, 5, (2, 3, 4))
pprint(tensor)
```

```python
torch.unflatten(tensor, -1, (2, 2))
```

`unflatten` расширяет измерение входного тензора на несколько измерений.
